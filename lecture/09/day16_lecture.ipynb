## Tufte's lie factor

<img src="images/day16_lie_factor_nyt.png" alt="A New York Times chart of required fuel economy: a road in perspective, with the line for 18 miles per gallon in 1978 labeled 0.6 inches long and the line for 27.5 miles per gallon in 1985 labeled 5.3 inches long" style="display:block;margin:0 auto;max-height:300px">

<div style="font-size:0.75em; text-align:center; color:#666;">New York Times, Aug 9 1978, p. D-2, as reproduced in Tufte 1983, <i>The Visual Display of Quantitative Information</i>.</div>

* lie factor = size of the effect in the graphic / size of the effect in the data
* in the data, 18 to 27.5 miles per gallon is a 53% increase. In the graphic, 0.6 to 5.3 inches is 783%
* 783 / 53 is about 15
* a tilted 3D pie does this too: the front slices look bigger

**Keep the lie factor near 1.**

```python
# Two bars with a movable baseline: the lie factor of the Hubble tension (a "skip" cell)
def plot_h0_bars(baseline):
    names = ['Planck 2018', 'SH0ES 2022']
    h0 = np.array([67.4, 73.0])            # km/s/Mpc, rounded
    fig, ax = plt.subplots(figsize=(6.2, 3.9))
    ax.bar(names, h0 - baseline, bottom=baseline, color=['C0', 'C1'], width=0.5)
    ax.set_ylim(baseline, 76)
    ax.set_ylabel('$H_0$ [km/s/Mpc]')
    drawn = (h0[1] - baseline) / (h0[0] - baseline) - 1
    true = h0[1] / h0[0] - 1
    ax.set_title(f'drawn difference {100 * drawn:.0f}%, true difference {100 * true:.0f}%\nlie factor = {drawn / true:.1f}', fontsize=13)
    plt.tight_layout()
    plt.show()
```

## Two bars, one slider

* $H_0$ is today's expansion rate. Planck, from the CMB, gets 67.4 km/s/Mpc, and SH0ES, from Cepheids and supernovae, gets 73.0
* that's an 8% difference, and it's called the Hubble tension
* move the baseline of the bars up. What happens to the lie factor?

<div style="font-size:0.75em; color:#666;">Planck Collaboration 2020, A&amp;A 641, A6 (67.4); Riess et al. 2022, ApJL 934, L7 (73.0). Values rounded.</div>

```python
# bars with a baseline slider
controls_right(plot_h0_bars, baseline=FloatSlider(value=0, min=0, max=66, step=1, description='baseline'))
```

* at a baseline of 0 the lie factor is 1
* at 60, note how the same two numbers look like a 76% difference. The lie factor is about 9!

**Start bars at zero, or don't use bars.**

```python
# SDSS stars with optional non-data ink, and an estimate of the data-ink ratio (a "skip" cell)
_sdss = np.load('../08/data/sdss_sspp_teff_logg.npz')
_pick = np.random.default_rng(16).choice(len(_sdss['teff']), 3000, replace=False)
_teff, _logg = _sdss['teff'][_pick], _sdss['logg'][_pick]

def _ink_fig(show_data, gray, grid, legend, frame, title):
    fig = Figure(figsize=(7, 4.6), dpi=80)
    FigureCanvasAgg(fig)
    ax = fig.add_subplot(111)
    ax.set_facecolor('0.85' if gray else 'white')
    ax.scatter(_teff, _logg, s=3, color='k', alpha=1.0 if show_data else 0.0)
    ax.invert_xaxis()
    ax.invert_yaxis()
    ax.set_xlabel('$T_{\\mathrm{eff}}$ [K]')
    ax.set_ylabel('$\\log g$')
    if grid: ax.grid(True, color='k', lw=1)
    if legend: ax.legend(handles=[Line2D([], [], marker='o', ls='', color='k', label='SDSS stars')], loc='upper left', frameon=True, shadow=True)
    if not frame:
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
    if title: ax.set_title('Stars in the SDSS SEGUE Stellar Parameter Pipeline')
    fig.tight_layout()
    return fig

def _ink(fig):
    fig.canvas.draw()
    pix = np.asarray(fig.canvas.buffer_rgba())[..., :3].astype(float) / 255
    return (1 - pix.mean(axis=2)).sum()

def plot_ink(gray, grid, legend, frame, title):
    with_data = _ink_fig(True, gray, grid, legend, frame, title)
    without = _ink_fig(False, gray, grid, legend, frame, title)
    ratio = (_ink(with_data) - _ink(without)) / _ink(with_data)
    display(with_data)
    print(f'data-ink ratio, our estimate: {ratio:.2f}')
```

## Tufte's data-ink ratio

* data-ink ratio = ink that shows data / all the ink used to print the figure
* these are SDSS stars again. Switch the extras off one at a time, and read the ratio printed under the plot. Which ones matter?
* the ratio is our estimate: (the ink with the stars, minus the ink without them) / the ink with the stars

```python
# SDSS stars with switchable non-data ink
controls_right(plot_ink, gray=Checkbox(value=True, description='gray panel'), grid=Checkbox(value=True, description='grid lines'), legend=Checkbox(value=True, description='legend box'), frame=Checkbox(value=True, description='top and right edges'), title=Checkbox(value=True, description='title'))
```

* everything on, about a quarter of the ink is data. Everything off, close to 90%
* the gray panel and the grid move the number most. The legend, the frame edges and the title barely register, but none of them is data
* the axes and labels stay, and we need them

**Delete the ink that isn't data, and keep the axes and labels.**

## Tufte's rules

1. The representation of numbers, as physically measured on the surface of the graph itself, should be directly proportional to the numerical quantities represented
2. Clear, detailed and thorough labeling should be used to defeat graphical distortion and ambiguity. Write out explanations of the data on the graph itself. Label important events in the data.
3. Show data variation, not design variation.
4. In time-series displays of money, deflated and standardized units of monetary measurement are nearly always better than nominal units.
5. The number of information carrying (variable) dimensions depicted should not exceed the number of dimensions in the data. Graphics must not quote data out of context.

**Rule 1 is the lie factor, rule 2 is labels, and rule 5 is Thursday's 3D pie.**

<div style="font-size:0.75em; text-align:center; color:#666;">Tufte 1983, <i>The Visual Display of Quantitative Information</i>.</div>

* the National Weather Service reflectivity map (top row) puts red next to green. Reflectivity is how strongly the rain scatters the radar beam. The bottom row is Chase Spectral
* same radar scan in all six panels: normal vision, then deuteranopia and protanopia, two kinds of red-green color-vision deficiency
* in the top row we lose the structure. In the bottom row lightness increases steadily, and the storm is still there. That's Crameri's point again

**Test a color map the way a color-blind person would see it.**

```python
# A simulated inclined rotating galaxy: a signed velocity field and a positive brightness map (a "skip" cell)
def _galaxy(n=180, inc_deg=60):
    x = np.linspace(-1, 1, n)
    X, Y = np.meshgrid(x, x)
    inc = np.deg2rad(inc_deg)
    Yd = Y / np.cos(inc)
    r = np.hypot(X, Yd)
    phi = np.arctan2(Yd, X)
    v = 220 * np.tanh(r / 0.2) * np.sin(inc) * np.cos(phi)          # km/s relative to the systemic velocity
    sb = np.exp(-r / 0.35)                                           # exponential disk
    off = r > 1
    return np.ma.masked_where(off, v), np.ma.masked_where(off, sb)

def plot_galaxy(field, cmap):
    v, sb = _galaxy()
    if field == 'velocity':
        data, label, kw = v, 'velocity relative to the center [km/s] (positive = receding)', dict(vmin=-np.abs(v).max(), vmax=np.abs(v).max())
    else:
        data, label, kw = sb, 'surface brightness [relative]', dict(vmin=0, vmax=1)
    fig, ax = plt.subplots(figsize=(5.4, 4.5))
    im = ax.imshow(data, origin='lower', cmap=cmap, extent=[-1, 1, -1, 1], **kw)
    fig.colorbar(im, ax=ax, label=label)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_ylim(-0.55, 0.55)
    plt.tight_layout()
    plt.show()
```

## The right color map for the task

* let's say we have one simulated galaxy and two tasks. Where is the velocity zero? Where is the brightness highest?
* a diverging map uses one color below a middle value, another above, and white in between. A sequential map is one color ramp (viridis or gray)
* pick a map for each field (i.e. velocity and brightness). Which map makes each answer easier to find?


```python
# a simulated rotating galaxy, two fields, four color maps
controls_right(plot_galaxy, field=Dropdown(options=['velocity', 'brightness'], description='field'), cmap=Dropdown(options=[('viridis (sequential)', 'viridis'), ('diverging', 'RdBu_r'), ('rainbow (jet)', 'jet'), ('gray (sequential)', 'gray')], description='color map'))
```

* the diverging map is white at zero, and the zero-velocity line is where the colors flip
* viridis has no special color at zero, and jet shows bands in both fields that aren't in the data
* brightness has no middle, and one color ramp is fine
* Borkin et al. 2011 asked 21 medical students to mark regions of low shear stress (the drag of blood on the artery wall) in arteries. That's finding where a value crosses a threshold, like our zero-velocity line
* a diverging map found about 47% more regions than a rainbow map in 2D, and 82% more in 3D

**Use a diverging map when the middle means something, and one ramp when it doesn't.**

* A (viridis) and C (gray) are perceptually uniform, so equal steps in brightness look like equal steps in color
* D is a rainbow map, like jet. It adds bands that aren't in the data
* B and E are one-hue maps. The EHT chose orange for E because orange looks hot (Ochigame et al. 2026). Some on the team argued for blue, because the hottest flame is blue
* is that acceptable? Depends on what you are trying to show. For finding where it's brightest, one ramp is the right kind, so A, B, C and E work and D doesn't

**The data are the same in all five panels. The color map is a choice you make.**

## DESI DR2: what would we change?

<img src="images/desi2025_fig11_w0wa_contours.png" alt="Contours in the w0-wa plane for DESI plus CMB combined with three supernova samples, Pantheon+, Union3 and DES Y5, with DESI plus CMB alone dashed; dashed gray lines mark w0 = -1 and wa = 0" style="display:block;margin:0 auto;max-height:250px">

<div style="font-size:0.75em; text-align:center; color:#666;">DESI Collaboration 2025, DESI DR2 Results II, arXiv:2503.14738, Fig. 11, CC BY 4.0.</div>

* DESI, the Dark Energy Spectroscopic Instrument, maps millions of galaxies to measure how the Universe expanded
* with the cosmic microwave background (CMB) and three supernova samples (Pantheon+, Union3, DES Y5 - my field), it constrains dark energy
* $w$ is dark energy's pressure over its density, and it can change as the Universe expands: $w_0$ is today's value and $w_a$ is how fast it changes
* a cosmological constant is the simplest dark energy, with $w_0 = -1$ and $w_a = 0$, where the gray dashed lines cross
* each contour encloses 68% or 95% of the posterior, the values of $w_0$ and $w_a$ that the data allow

## Wrapping up communication

You should be able to check a figure for:

* ambiguity: is what went into the plot labeled on the plot?
* distortion: do the bars start at zero, and is the lie factor near 1?
* distraction: how much of the ink shows data?
* small multiples: the same axes in every panel
* color: the right map for the task, grayscale and color-blind people, and how we stretched an image
* bands: the model we're testing against, with room around it, and what the band contains

**Every plot is designed to tell a story. Make sure other people can check it.**
